```python
cat_feats = ['purpose']
```

```python
final_data = pd.get_dummies(loans, columns=cat_feats, drop_first=True, dtype=int)
```

```python
final_data.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = final_data.drop('not.fully.paid', axis=1)
y = final_data['not.fully.paid']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree = DecisionTreeClassifier(random_state=42)
```

```python
dtree.fit(X_train, y_train)
```

```python
predictions = dtree.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, predictions, zero_division=0))
```

```python
print(confusion_matrix(y_test, predictions))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
rfc.fit(X_train, y_train)
```

```python
rfc_pred = rfc.predict(X_test)
```

```python
print(classification_report(y_test, rfc_pred, zero_division=0))
```

```python
# Optional: compare accuracy and F1 of the positive class
from sklearn.metrics import accuracy_score, f1_score
print('Decision Tree Accuracy:', accuracy_score(y_test, predictions))
print('Random Forest Accuracy:', accuracy_score(y_test, rfc_pred))
print('Decision Tree F1 (not fully paid):', f1_score(y_test, predictions, zero_division=0))
print('Random Forest F1 (not fully paid):', f1_score(y_test, rfc_pred, zero_division=0))
```

```python
print(confusion_matrix(y_test, rfc_pred))
```

**What performed better the random forest or the decision tree?**

The Random Forest performed better in terms of overall accuracy (83.61%) compared to the Decision Tree (73.52%). However, the Decision Tree achieved a higher F1-score for class 1 (20.48% vs. 2.89%), making it better at identifying loans that were not fully paid.

Therefore, Random Forest is better for overall accuracy, while Decision Tree is better for detecting class 1.
