Display Mean-Imputed Dataset

Display Median-Imputed Dataset

```python
plt.figure(figsize=(4, 3))

sns.boxplot(x=df['final_exam_score'])

plt.title("Boxplot of Final Exam Score")
plt.show()
```

Boxplot

```python
Q1 = df['final_exam_score'].quantile(0.25)
Q3 = df['final_exam_score'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['final_exam_score'] < lower) |
    (df['final_exam_score'] > upper)
]

outliers.head(15)
```

Detect Outliers Using IQR

```python
df_no_outliers = df[
    (df['final_exam_score'] >= lower) &
    (df['final_exam_score'] <= upper)
]

print("Original shape: ", df.shape)
print("After removing outliers: ", df_no_outliers.shape)
```

Remove Outliers

```python
lower_cap = df['final_exam_score'].quantile(0.05)
upper_cap = df['final_exam_score'].quantile(0.95)

df_capped = df.copy()

df_capped['final_exam_score'] = (
    df_capped['final_exam_score'].clip(
        lower_cap,
        upper_cap
    )
)
```

Capping Outliers

```python
df[['final_exam_score', 'attendance_percentage']].head()
```

Display Selected Numerical Features

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_scaled = df[
    ['final_exam_score', 'attendance_percentage']
].copy()

df_scaled[
    ['final_exam_score', 'attendance_percentage']
] = scaler.fit_transform(df_scaled)

df_scaled.head()
```

Min-Max Normalization

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_standardized = df[
    ['final_exam_score', 'attendance_percentage']
].copy()

df_standardized[
    ['final_exam_score', 'attendance_percentage']
] = scaler.fit_transform(df_standardized)

df_standardized.head()
```

Z-Score Standardization

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    df_standardized[
        ['final_exam_score', 'attendance_percentage']
    ].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap (Before PCA)")
plt.show()
```

Correlation Heatmap

```python
# Import PCA (Principal Component Analysis) from scikit-learn
from sklearn.decomposition import PCA

# Select the standardized features to be used for PCA
X = df_standardized[
    ['final_exam_score', 'attendance_percentage']
]

# Create a PCA model with 2 principal components
pca = PCA(n_components=2)

# Apply PCA to the selected data
principal_components = pca.fit_transform(X)

# Display the proportion of variance explained
print(
    "Explained Variance Ratio:",
    pca.explained_variance_ratio_
)
```

Apply PCA

```python
plt.figure(figsize=(6, 4))

plt.scatter(
    principal_components[:, 0],
    principal_components[:, 1]
)

plt.title("PCA Projection")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```
