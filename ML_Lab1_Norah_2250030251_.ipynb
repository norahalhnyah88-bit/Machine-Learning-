```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

for mark in marks:
    if mark >= 60:
        print(mark, "Pass")
    else:
        print(mark, "Fail")

print("\nStudy Hours, Mark, and Result:")

for i in range(len(study_hours)):
    if marks[i] >= 60:
        result = "Pass"
    else:
        result = "Fail"

    print("Study hours:", study_hours[i],
          "Mark:", marks[i],
          "Result:", result)
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]
# Dataset
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]
# Your task: complete the analysis below
# 1. Number of samples
# 2. Average study hours
# 3. Average marks
# 4. Highest and lowest marks
# 5. Display each sample
# 6. Pass/Fail
# 7. Count the number of passes


# 1. Number of samples
number_of_samples = len(marks)
print("Number of samples:", number_of_samples)

# 2. Average study hours
average_study_hours = sum(study_hours) / len(study_hours)
print("Average study hours:", average_study_hours)

# 3. Average marks
average_marks = sum(marks) / len(marks)
print("Average mark:", average_marks)

# 4. Highest and lowest marks
highest_mark = max(marks)
lowest_mark = min(marks)
print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)

# 5. Display each study-hours/mark pair
print("\nStudy Hours and Marks:")

for i in range(len(study_hours)):
    print("Study hours:", study_hours[i], "Mark:", marks[i])

# 6. Display Pass or Fail for each sample
print("\nPass or Fail:")

for i in range(len(marks)):
    if marks[i] >= 60:
        print("Study hours:", study_hours[i],
              "Mark:", marks[i],
              "Result: Pass")
    else:
        print("Study hours:", study_hours[i],
              "Mark:", marks[i],
              "Result: Fail")

# 7. Count the number of passes
pass_count = 0

for mark in marks:
    if mark >= 60:
        pass_count = pass_count + 1

print("\nNumber of students who passed:", pass_count)
```
