```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('student_performance.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['final_exam_score'], bins=10, edgecolor='black')

plt.title('Distribution of Final Exam Scores')
plt.xlabel('Final Exam Score')
plt.ylabel('Frequency')

plt.show()
```

Observation: Most students' final exam scores are concentrated between 55 and 85, with the highest frequency around 70–75.

```python
sns.boxplot(x=df['final_exam_score'])

plt.title('Box Plot of Final Exam Scores')
plt.xlabel('Final Exam Score')

plt.show()
```

Observation: The median final exam score is around 68. Most scores are between approximately 57 and 79, and no clear outliers are observed.

```python
sns.countplot(x='pass_fail', data=df)

plt.title('Pass vs Fail')
plt.xlabel('Result (0 = Fail, 1 = Pass)')
plt.ylabel('Number of Students')

plt.show()
```

Observation: Most students passed. Approximately 83% of the students passed, while 17% failed.

```python
sns.scatterplot(
    x='study_hours_per_week',
    y='final_exam_score',
    data=df
)

plt.title('Study Hours vs Final Exam Score')
plt.xlabel('Study Hours per Week')
plt.ylabel('Final Exam Score')

plt.show()
```

Observation: There is a positive relationship between study hours per week and final exam scores. Students who study more hours tend to achieve higher exam scores.

```python
df['study_hours_per_week'].corr(df['final_exam_score'])
```

Observation: The correlation between study hours per week and final exam scores is approximately 0.73, indicating a strong positive relationship between the two variables.

```python
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(10, 8))

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap')

plt.show()
```

Observation: The heatmap shows several strong positive correlations among academic performance variables. Test 1 and Test 2 scores have a strong positive correlation (0.83). Final exam scores are also positively correlated with Test 1 (0.79), Test 2 (0.78), and study hours per week (0.73). Attendance and assignments submitted also show a strong positive correlation (0.79).

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x='subject', y='final_exam_score', data=df)

plt.title('Final Exam Scores by Subject')
plt.xlabel('Subject')
plt.ylabel('Final Exam Score')

plt.show()
```

Observation: Final exam scores are relatively similar across subjects. Art, English, and Computer have slightly higher median scores, but there are no major differences between the subjects.

```python
sns.scatterplot(
    x='attendance_percentage',
    y='final_exam_score',
    data=df
)

plt.title('Attendance vs Final Exam Score')
plt.xlabel('Attendance Percentage')
plt.ylabel('Final Exam Score')

plt.show()
```

Observation: There is a positive relationship between attendance percentage and final exam scores. Students with higher attendance tend to achieve higher exam scores.

```python
sns.countplot(
    x='final_grade_letter',
    data=df,
    order=['A', 'B', 'C', 'D', 'F']
)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Number of Students')

plt.show()
```

Observation: Grade C is the most common final grade, while grade A is the least common. Most students received grades between C and F.

The exploratory data analysis showed several patterns in student performance. Study hours and attendance have positive relationships with final exam scores. Test 1 and Test 2 scores are also strongly correlated with final exam performance. Most students passed, and grade C was the most common final grade. Overall, the analysis suggests that study habits, attendance, and previous test performance are associated with students' academic performance.
